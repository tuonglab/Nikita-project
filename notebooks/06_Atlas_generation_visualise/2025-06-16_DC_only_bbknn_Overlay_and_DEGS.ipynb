```python
# set colours for bar graph
coloursDC = data_DC.uns["NR_annotations_simple_colors"]
clusters = data_DC.obs["NR_annotations_simple"].cat.categories
colour_dict = dict(zip(clusters, coloursDC))
```

```python
#plot stacked graph by sample type
ct = pd.crosstab(data_DC.obs["sample_type_major2"], data_DC.obs["NR_annotations_simple"])
ct_pct = ct.apply(lambda r: r / r.sum() * 100, axis=1)

# set order for stack
#ct_pct = ct_pct[["DC3-like cDC2","classical cDC2","ISG cDC2","SIK3 cDC2","CD207-S100 cDC2","CD207-LTB cDC2","CD207-IL18 cDC2"]]

#set order for sample types 
ct_pct = ct_pct.loc[["healthy","primary tumour","metastatic tumour","ascites"]]

colours = [colour_dict[col] for col in ct_pct.columns]
ct_pct.plot.bar(stacked=True, figsize=(4,7), color=colours)
plt.legend(bbox_to_anchor=(1, 0.5), loc="center left", frameon=False)
plt.title("DC Clusters by Sample Type")
plt.tight_layout()  
plt.savefig("barplot_DC_sample_type.pdf", format="pdf", bbox_inches="tight")
plt.show()
```
